### 🤝 We Do: Compound Interest Calculator

**Scenario:** You invest **Ksh 1,000** in a savings account with a **10% annual return**.  
How much do you have after **7 years**? After **15 years**? After **30 years**? 💰

````python
# 💰 THE MAGIC OF COMPOUND INTEREST
principal = 1_000   # Starting amount (Ksh)
rate = 0.10         # 10% annual return

# Formula: Final = Principal × (1 + rate)^years
after_1_year  = principal * (1 + rate) ** 1
after_7_years = principal * (1 + rate) ** 7
after_15_years = principal * (1 + rate) ** 15
after_30_years = principal * (1 + rate) ** 30

print("💰 COMPOUND INTEREST GROWTH")
print(f"   Starting principal: Ksh {principal:,.0f}")
print(f"   Annual rate       : {rate*100:.0f}%")
print()
print(f"   📅 After  1 year  : Ksh {after_1_year:>10,.2f}")
print(f"   📅 After  7 years : Ksh {after_7_years:>10,.2f}")
print(f"   📅 After 15 years : Ksh {after_15_years:>10,.2f}")
print(f"   📅 After 30 years : Ksh {after_30_years:>10,.2f}")
print()
print(f"   🚀 In 30 years, your Ksh 1,000 grows {after_30_years/principal:.1f}x!")
print("   This is why starting early matters! 🌱")
````

## 4️⃣ Control Flow — Making Decisions 🚦

Control flow lets your code **think** and **decide**. Just like in real life!

```
🌧️ IF it's raining heavily → bring an umbrella
☁️ ELIF it's cloudy        → bring a light jacket
☀️ ELSE                    → enjoy the sunshine!
```

### The Structure:
```python
if <condition>:
    # runs when condition is True
elif <another condition>:
    # runs when this condition is True
else:
    # runs when ALL conditions above are False
```

### Comparison Operators:
| Operator | Meaning | Example |
|----------|---------|--------|
| `==` | Equal to | `city == "Nairobi"` |

## 5️⃣ Loops — Repeat After Me! 🔄

Loops let you **automate repetition** — no more copy-pasting!

### Two Types:

```python
# 🔁 FOR loop — when you know HOW MANY times to repeat
for item in collection:
    do_something(item)

# 🔂 WHILE loop — when you repeat UNTIL a condition is False
while condition_is_true:
    do_something()
```

> 🌍 **Real world analogy:**  
> `for` → "Check every student's exam score"  
> `while` → "Keep selling until stock runs out"

### 👀 I Do: For Loops

### 👀 I Do: While Loops

````python
# 🔂 WHILE LOOP — Simulate water tank usage
total_water = 1000   # litres
daily_usage = 150    # litres per day
day = 1

print("💧 WATER TANK SIMULATION")
print(f"   Starting water: {total_water}L")
print(f"   Daily usage   : {daily_usage}L")
print("-" * 35)

while total_water > 0:
    total_water -= daily_usage

    if total_water >= 0:
        bar = "💧" * (total_water // 100)
        print(f"  Day {day:>2}: {max(total_water,0):>5}L remaining  {bar}")
    else:
        print(f"  Day {day:>2}: Tank ran out mid-day!")

    day += 1

print("-" * 35)
print(f"🚰 Tank is EMPTY after {day-1} days! Time to refill!")
````

3. Track `total_earnings` and `day_count`
4. Stop once earnings ≥ 5,000 or all 7 days are done
5. Print: which day the target was hit (if at all) and total earnings

````python
# 🌽 EXERCISE 1: Maize Farmer Sales Simulator
daily_harvest = [100, 120, 80, 90, 110, 95, 105]  # kg per day
price_per_kg = 50                                   # Ksh
earnings_goal = 5_000                               # Ksh

# Your code here 👇
total_earnings = 0
day = 0

# Hint: while day < len(daily_harvest) and total_earnings < earnings_goal:

````

````python
# ✅ EXERCISE 1 — SOLUTION
daily_harvest = [100, 120, 80, 90, 110, 95, 105]
price_per_kg = 50
earnings_goal = 5_000

total_earnings = 0
day = 0
goal_reached_day = None

print("🌽 MAIZE SALES SIMULATOR — Kitale Farmer")
print("=" * 50)

while day < len(daily_harvest) and total_earnings < earnings_goal:
    harvest = daily_harvest[day]
    earnings_today = harvest * price_per_kg
    total_earnings += earnings_today
    day += 1

    status = "✅ GOAL HIT!" if total_earnings >= earnings_goal and goal_reached_day is None else ""
    if total_earnings >= earnings_goal and goal_reached_day is None:
        goal_reached_day = day

    print(f"  Day {day}: {harvest}kg × Ksh{price_per_kg} = Ksh{earnings_today:,}  "
          f"| Total: Ksh{total_earnings:,}  {status}")

print("=" * 50)
if goal_reached_day:
    print(f"🎉 Goal of Ksh {earnings_goal:,} reached on Day {goal_reached_day}!")
else:
    print(f"😔 Goal not reached. Total after 7 days: Ksh {total_earnings:,}")
````
